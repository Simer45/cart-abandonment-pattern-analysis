```python
import pandas as pd

events = pd.read_csv("C:/Projects/Cart-Abandoment-Pattern-Analysis/Data/events.csv")

print(events.shape)
print(events.dtypes)
events.head()
```

```python
print("Event type counts:\n", events['event'].value_counts())
print("\nNulls:\n", events.isnull().sum())
print("\nDuplicate rows:", events.duplicated().sum())

# Convert timestamp (milliseconds since 1970) into a real datetime
events['event_time'] = pd.to_datetime(events['timestamp'], unit='ms')
print("\nDate range:", events['event_time'].min(), "to", events['event_time'].max())

print("\nUnique visitors:", events['visitorid'].nunique())
print("Unique items:", events['itemid'].nunique())
```

```python
print("Before cleaning:", events.shape)

# Drop exact duplicates
events = events.drop_duplicates()

print("After cleaning:", events.shape)

# Keep only the funnel-relevant events for the main analysis table
funnel = events[events['event'].isin(['addtocart', 'transaction'])].copy()
print("Funnel-relevant rows:", funnel.shape)

output_path = "C:/Projects/Cart-Abandoment-Pattern-Analysis/exports/funnel_events_clean.csv"
funnel.to_csv(output_path, index=False)
print("Saved:", output_path)
```
